```python
#train - validation split
import os
import shutil
import random
# Go to project root (one level up from notebooks)
os.chdir("..")

print("Current working directory:", os.getcwd())
# Paths
image_dir = "data/processed/images/all"
label_dir = "data/processed/labels/all"

train_img_dir = "data/processed/images/train"
val_img_dir = "data/processed/images/val"

train_lbl_dir = "data/processed/labels/train"
val_lbl_dir = "data/processed/labels/val"

# Create folders
for path in [train_img_dir, val_img_dir, train_lbl_dir, val_lbl_dir]:
    os.makedirs(path, exist_ok=True)

# Get all images
images = os.listdir(image_dir)
random.shuffle(images)

# Split (80-20)
split = int(0.8 * len(images))
train_images = images[:split]
val_images = images[split:]

# Move files
def move_files(image_list, img_dest, lbl_dest):
    for img_name in image_list:
        label_name = img_name.replace(".jpg", ".txt")
        
        shutil.copy(os.path.join(image_dir, img_name), os.path.join(img_dest, img_name))
        shutil.copy(os.path.join(label_dir, label_name), os.path.join(lbl_dest, label_name))

move_files(train_images, train_img_dir, train_lbl_dir)
move_files(val_images, val_img_dir, val_lbl_dir)

print("✅ Train/Val Split Done!")
```
